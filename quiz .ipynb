```python
# General Knowledge Quiz
# DecodeLabs Project 4

questions = [
    {
        "question": "What is the capital of Pakistan?",
        "answer": "islamabad",
        "category": "Geography"
    },
    {
        "question": "Who developed Python?",
        "answer": "guido van rossum",
        "category": "Programming"
    },
    {
        "question": "Which planet is known as the Red Planet?",
        "answer": "mars",
        "category": "Science"
    },
    {
        "question": "What is the largest ocean in the world?",
        "answer": "pacific ocean",
        "category": "Geography"
    },
    {
        "question": "How many continents are there on Earth?",
        "answer": "7",
        "category": "General Knowledge"
    }
]

while True:

    print("\n" + "=" * 50)
    print("      ADVANCED GENERAL KNOWLEDGE QUIZ")
    print("=" * 50)

    username = input("Enter your name: ").strip().title()

    score = 0

    for index, q in enumerate(questions, start=1):

        print(f"\nQuestion {index}")
        print(f"Category: {q['category']}")
        print(q["question"])

        user_answer = input("Your Answer: ").strip().lower()

        if user_answer == q["answer"]:
            print("Correct!")
            score += 1
        else:
            print(f"Wrong!")
            print(f"Correct Answer: {q['answer'].title()}")

    percentage = (score / len(questions)) * 100

    print("\n" + "=" * 50)
    print("RESULT")
    print("=" * 50)

    print(f"Name: {username}")
    print(f"Score: {score}/{len(questions)}")
    print(f"Percentage: {percentage:.2f}%")

    if percentage >= 80:
        grade = "A"
        remark = "Excellent Performance!"
    elif percentage >= 60:
        grade = "B"
        remark = "Good Job!"
    elif percentage >= 40:
        grade = "C"
        remark = "Keep Practicing!"
    else:
        grade = "F"
        remark = "Needs Improvement!"

    print(f"Grade: {grade}")
    print(f"Remark: {remark}")

    play_again = input(
        "\nDo you want to play again? (yes/no): "
    ).strip().lower()

    if play_again != "yes":
        print("\nThank you for using the Quiz System!")
        break
```
